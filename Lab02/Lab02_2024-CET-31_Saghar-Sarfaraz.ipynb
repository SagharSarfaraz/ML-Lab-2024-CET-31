Q3:Seed 2026: one rand value, 25 randn values. Seed 42: roll 10,000 Ludo dice, count each face with np.unique, estimate P(six), count double sixes without a loop.

```python
import numpy as np

# Part A: Seed 2026
np.random.seed(2026)
# 1. Generate one random value
random_value = np.random.rand()
print("T3a: Random Value:", random_value)
```

```python
# 2. Generate 25 random normal values
random_normal = np.random.randn(25)
print("T3b: Random Normal Values:", random_normal)
```

```python
# Part B: Seed 42
np.random.seed(42)

# 3. Roll a Ludo dice 10,000 times
dice_rolls = np.random.randint(1, 7, 10000)
print("T3c: Total Dice Rolls:", len(dice_rolls), "(rolls)")
```

```python
# 4. Count each face using np.unique
faces, counts = np.unique(dice_rolls, return_counts=True)

print("T3d: Dice Face Counts:")
for face, count in zip(faces, counts):
    print(face, ":", count, "times")
```

```python
# 5. Estimate probability of getting six
six_count = np.sum(dice_rolls == 6)
probability_six = six_count / 10000

print("T3e: Probability of Six:", probability_six)
```

Q9:+5 capped at 100 with np.clip; pass %; who passed only because of grace; z-scores.

```python
import numpy as np

marks = np.array([45, 38, 32, 48, 55])
# A: Add 5 Grace Marks

new_marks = marks + 5

print("T9A:", new_marks)
```

```python
# B: Cap Marks at 100

new_marks = np.clip(new_marks, 0, 100)

print("T9B:", new_marks)
```

```python
# C: Pass Percentage

passed = np.sum(new_marks >= 40)

pass_percentage = (passed / len(new_marks)) * 100

print("T9C:", pass_percentage)
```

```python
# D: Z-Scores

z_scores = (new_marks - np.mean(new_marks)) / np.std(new_marks)

print("T9D:", z_scores)
```

```python
# E: Students Who Passed Only Because of Grace Marks

grace_passed = np.where((marks < 40) & (new_marks >= 40))

print("T9E:", grace_passed)
```
