## Task 3 · Clean the data: the six types of problem
Fix the problems **in this order** (text first, then duplicates):
- **(a) Missing values:** count them per column. Drop ads with no `Brand` or no `Model`; fill missing `Registered City`, `Fuel` and `Transaction Type` with `"Unknown"`; leave `Year` and `KMs Driven` as NaN. Print how many rows were removed.
- **(b) Formatting:** remove extra spaces and fix the case of city names (`.str.strip().str.title()`); remove `"Rs"` and commas from `Price`.
- **(c) Data types:** convert `Price` to `int64` and `Year` to `Int64`.
- **(d) Inconsistent categories:** `"Sukkar"` → `"Sukkur"` and `"Corrolla Altis"` → `"Corolla Altis"`. Print the number of different cities before and after.
- **(e) Duplicates:** count exact duplicate ads, drop them, print the new row count.
- **(f) Invalid values:** set `KMs Driven` to NaN when it is below 100, above 1,000,000, or the placeholder 12345 / 123456. Drop ads with the fake prices 12,345,678 or 87,654,321, and Suzuki/Daihatsu ads above 5,000,000.
- **(g) Self-check:** your final shape must be **(20484, 9)**.

```python
# Task 3: Clean the data in the required order

# T3a: Count missing values, remove rows without Brand or Model,
# and fill the specified text columns with "Unknown".
print("T3a: Missing values before cleaning:")
print(df.isna().sum())

rows_before_missing = len(df)
df = df.dropna(subset=["Brand", "Model"])
rows_removed_missing = rows_before_missing - len(df)

df[["Registered City", "Fuel", "Transaction Type"]] = (
    df[["Registered City", "Fuel", "Transaction Type"]].fillna("Unknown")
)
print("T3a: Rows removed because Brand or Model was missing:", rows_removed_missing)

# T3b: Fix city formatting and remove Rs and commas from Price.
df["Registered City"] = (
    df["Registered City"].astype("string").str.strip().str.title()
)

df["Price"] = (
    df["Price"].astype("string")
      .str.replace("Rs", "", regex=False)
      .str.replace(",", "", regex=False)
      .str.strip()
)
print("T3b: Sample cleaned prices:")
print(df["Price"].head(5))

# T3c: Convert Price and Year to the required data types.
df["Price"] = pd.to_numeric(df["Price"], errors="raise").astype("int64")
df["Year"] = df["Year"].astype("Int64")
print("T3c: Price dtype:", df["Price"].dtype)
print("T3c: Year dtype:", df["Year"].dtype)

# T3d: Correct inconsistent city and model categories.
cities_before = df["Registered City"].nunique()

df["Registered City"] = df["Registered City"].replace({"Sukkar": "Sukkur"})
df["Model"] = df["Model"].replace({"Corrolla Altis": "Corolla Altis"})

cities_after = df["Registered City"].nunique()
print("T3d: Different cities before correction:", cities_before)
print("T3d: Different cities after correction:", cities_after)

# T3e: Count and remove exact duplicate ads.
duplicate_count = df.duplicated().sum()
df = df.drop_duplicates()
print("T3e: Exact duplicate ads removed:", duplicate_count)
print("T3e: Rows after removing duplicates:", len(df))

# T3f: Replace invalid kilometre values with NaN.
invalid_km = (
    (df["KMs Driven"] < 100)
    | (df["KMs Driven"] > 1_000_000)
    | (df["KMs Driven"].isin([12345, 123456]))
)
invalid_km_count = invalid_km.sum()
df.loc[invalid_km, "KMs Driven"] = np.nan
print("T3f: Invalid kilometre values changed to NaN:", invalid_km_count)

# Remove fake prices and Suzuki/Daihatsu ads priced above Rs 5,000,000.
fake_price = df["Price"].isin([12_345_678, 87_654_321])
high_suzuki_daihatsu = (
    df["Brand"].isin(["Suzuki", "Daihatsu"]) & (df["Price"] > 5_000_000)
)
invalid_price_rows = fake_price | high_suzuki_daihatsu
invalid_price_count = invalid_price_rows.sum()
df = df.loc[~invalid_price_rows].copy()
print("T3f: Ads removed by invalid-price rules:", invalid_price_count)

# T3g: Required final self-check.
print("T3g: Final shape:", df.shape)
```

## Task 4 · Analyse
- **(a)** New column `Price_lakh` = price in lakh rupees (÷ 100,000).
- **(b)** New column `Age` = 2020 − `Year`.
- **(c)** The 5 most common brands.
- **(d)** The 5 cities with the most ads.
- **(e)** Median `Price_lakh` for each fuel type, highest first.
- **(f)** Median `Price_lakh` of a **Corolla GLI** for each year 2010–2018. How much value does it lose per year?
- **(g)** The 5 most expensive ads. Does one of them look wrong? Why?
- **(h)** Number of ads and median price for Cash vs Installment/Leasing. Interpret in 1–2 lines.

```python
# Task 4: Analyse the cleaned data

# T4a: Price in lakh rupees.
df["Price_lakh"] = df["Price"] / 100_000
print("T4a: Price_lakh added.")

# T4b: Age using the lab's specified reference year.
df["Age"] = 2020 - df["Year"]
print("T4b: Age added.")

# T4c: Five most common brands.
print("T4c: Five most common brands:")
print(df["Brand"].value_counts().head(5))

# T4d: Five cities with the most ads.
print("T4d: Five cities with the most ads:")
print(df["Registered City"].value_counts().head(5))

# T4e: Median price in lakh for each fuel type, highest first.
print("T4e: Median Price_lakh by fuel type:")
print(df.groupby("Fuel")["Price_lakh"].median().sort_values(ascending=False))

# T4f: Median Corolla GLI price for each year from 2010 through 2018.
gli_yearly_median = (
    df.loc[
        (df["Model"] == "Corolla GLI") & df["Year"].between(2010, 2018)
    ]
    .groupby("Year")["Price_lakh"]
    .median()
    .reindex(range(2010, 2019))
)
print("T4f: Median Corolla GLI Price_lakh by year:")
print(gli_yearly_median)

# Estimate the average yearly change from the 2010 and 2018 medians.
annual_change = (
    gli_yearly_median.loc[2018] - gli_yearly_median.loc[2010]
) / (2018 - 2010)
print("T4f: Average change per model year (lakh rupees):", round(annual_change, 2))
print("T4f interpretation: Newer model years have higher median prices;")
print("the median price falls by about", round(annual_change, 2),
      "lakh rupees for each additional year of age, based on these endpoints.")

# T4g: Five most expensive ads.
print("T4g: Five most expensive ads:")
print(df.nlargest(5, "Price")[["Brand", "Model", "Year", "Price", "Registered City", "Fuel"]])
print("T4g interpretation: The highest listing is expensive, but price alone cannot prove it is fake;")
print("it should be checked against the vehicle model, year, and comparable listings.")

# T4h: Compare Cash with Installment/Leasing only.
cash_installment = df.loc[
    df["Transaction Type"].isin(["Cash", "Installment/Leasing"])
]
print("T4h: Ad count and median price by transaction type:")
print(
    cash_installment.groupby("Transaction Type").agg(
        Ads=("Price", "size"),
        Median_Price_lakh=("Price_lakh", "median")
    )
)
print("T4h interpretation: Cash ads are much more numerous and have a higher median listed price")
print("than Installment/Leasing ads in this cleaned dataset.")
```
